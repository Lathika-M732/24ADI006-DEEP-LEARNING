# TASK A — Visualization of Activation Functions

```python
import numpy as np
import matplotlib.pyplot as plt

name = "Lathika M"
roll_no = "24BAD062"

print("EXPT NO: 2")
print("COMPARATIVE STUDY OF ACTIVATION FUNCTIONS")
print("Name:", name)
print("Roll No:", roll_no)
print()

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 1000)

sigmoid_output = sigmoid(x)
tanh_output = tanh(x)
relu_output = relu(x)

plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid_output, label="Sigmoid")
plt.plot(x, tanh_output, label="Tanh")
plt.plot(x, relu_output, label="ReLU")

plt.title("Comparison of Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid()
plt.show()

print("Activation Function Details")
print()
print("Sigmoid : Output Range = 0 to 1")
print("Tanh    : Output Range = -1 to 1")
print("ReLU    : Output Range = 0 to infinity")
```

# Task B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models

name = "Lathika M"
roll_no = "24BAD062"

print("EXPT NO: 2")
print("PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS")
print("Name:", name)
print("Roll No:", roll_no)
print()

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

def create_model(activation):
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation=activation),
        layers.Dense(64, activation=activation),
        layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model

sigmoid_model = create_model("sigmoid")

print("Training Sigmoid Model...")
history_sigmoid = sigmoid_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

tanh_model = create_model("tanh")

print("\nTraining Tanh Model...")
history_tanh = tanh_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

relu_model = create_model("relu")

print("\nTraining ReLU Model...")
history_relu = relu_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

sigmoid_test_loss, sigmoid_test_accuracy = sigmoid_model.evaluate(
    x_test, y_test, verbose=0
)

tanh_test_loss, tanh_test_accuracy = tanh_model.evaluate(
    x_test, y_test, verbose=0
)

relu_test_loss, relu_test_accuracy = relu_model.evaluate(
    x_test, y_test, verbose=0
)

print("\nFINAL RESULTS")
print("=" * 50)

print("Sigmoid")
print("Training Accuracy:", history_sigmoid.history["accuracy"][-1])
print("Validation Accuracy:", history_sigmoid.history["val_accuracy"][-1])
print("Training Loss:", history_sigmoid.history["loss"][-1])
print("Validation Loss:", history_sigmoid.history["val_loss"][-1])
print("Test Accuracy:", sigmoid_test_accuracy)
print("Test Loss:", sigmoid_test_loss)

print("\nTanh")
print("Training Accuracy:", history_tanh.history["accuracy"][-1])
print("Validation Accuracy:", history_tanh.history["val_accuracy"][-1])
print("Training Loss:", history_tanh.history["loss"][-1])
print("Validation Loss:", history_tanh.history["val_loss"][-1])
print("Test Accuracy:", tanh_test_accuracy)
print("Test Loss:", tanh_test_loss)

print("\nReLU")
print("Training Accuracy:", history_relu.history["accuracy"][-1])
print("Validation Accuracy:", history_relu.history["val_accuracy"][-1])
print("Training Loss:", history_relu.history["loss"][-1])
print("Validation Loss:", history_relu.history["val_loss"][-1])
print("Test Accuracy:", relu_test_accuracy)
print("Test Loss:", relu_test_loss)

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["accuracy"],
    label="Sigmoid Training"
)

plt.plot(
    history_sigmoid.history["val_accuracy"],
    label="Sigmoid Validation"
)

plt.plot(
    history_tanh.history["accuracy"],
    label="Tanh Training"
)

plt.plot(
    history_tanh.history["val_accuracy"],
    label="Tanh Validation"
)

plt.plot(
    history_relu.history["accuracy"],
    label="ReLU Training"
)

plt.plot(
    history_relu.history["val_accuracy"],
    label="ReLU Validation"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["loss"],
    label="Sigmoid Training"
)

plt.plot(
    history_sigmoid.history["val_loss"],
    label="Sigmoid Validation"
)

plt.plot(
    history_tanh.history["loss"],
    label="Tanh Training"
)

plt.plot(
    history_tanh.history["val_loss"],
    label="Tanh Validation"
)

plt.plot(
    history_relu.history["loss"],
    label="ReLU Training"
)

plt.plot(
    history_relu.history["val_loss"],
    label="ReLU Validation"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()

print("\nBEST VALIDATION ACCURACY")

sigmoid_best = max(history_sigmoid.history["val_accuracy"])
tanh_best = max(history_tanh.history["val_accuracy"])
relu_best = max(history_relu.history["val_accuracy"])

print("Sigmoid:", sigmoid_best)
print("Tanh:", tanh_best)
print("ReLU:", relu_best)

best_activation = max(
    [
        ("Sigmoid", sigmoid_best),
        ("Tanh", tanh_best),
        ("ReLU", relu_best)
    ],
    key=lambda x: x[1]
)

print("\nBest Activation Function:", best_activation[0])
print("Best Validation Accuracy:", best_activation[1])
```

# Task C: COMPARISON OF OPTIMIZATION ALGORITHMS

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models

name = "Lathika M"
roll_no = "24BAD062"

print("EXPT NO: 2")
print("COMPARISON OF OPTIMIZATION ALGORITHMS")
print("Name:", name)
print("Roll No:", roll_no)
print()

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

def create_model(optimizer):
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation="relu"),
        layers.Dense(64, activation="relu"),
        layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model

sgd_model = create_model(
    tf.keras.optimizers.SGD(learning_rate=0.01)
)

print("Training SGD Model...")

history_sgd = sgd_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

momentum_model = create_model(
    tf.keras.optimizers.SGD(
        learning_rate=0.01,
        momentum=0.9
    )
)

print("\nTraining Momentum Model...")

history_momentum = momentum_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

rmsprop_model = create_model(
    tf.keras.optimizers.RMSprop(
        learning_rate=0.001
    )
)

print("\nTraining RMSProp Model...")

history_rmsprop = rmsprop_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

adam_model = create_model(
    tf.keras.optimizers.Adam(
        learning_rate=0.001
    )
)

print("\nTraining Adam Model...")

history_adam = adam_model.fit(
    x_train,
    y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

sgd_test_loss, sgd_test_accuracy = sgd_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

momentum_test_loss, momentum_test_accuracy = momentum_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

rmsprop_test_loss, rmsprop_test_accuracy = rmsprop_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

adam_test_loss, adam_test_accuracy = adam_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nFINAL RESULTS")
print("=" * 60)

print("\nSGD")
print("Training Accuracy:",
      history_sgd.history["accuracy"][-1])
print("Validation Accuracy:",
      history_sgd.history["val_accuracy"][-1])
print("Training Loss:",
      history_sgd.history["loss"][-1])
print("Validation Loss:",
      history_sgd.history["val_loss"][-1])
print("Test Accuracy:",
      sgd_test_accuracy)
print("Test Loss:",
      sgd_test_loss)

print("\nMomentum")
print("Training Accuracy:",
      history_momentum.history["accuracy"][-1])
print("Validation Accuracy:",
      history_momentum.history["val_accuracy"][-1])
print("Training Loss:",
      history_momentum.history["loss"][-1])
print("Validation Loss:",
      history_momentum.history["val_loss"][-1])
print("Test Accuracy:",
      momentum_test_accuracy)
print("Test Loss:",
      momentum_test_loss)

print("\nRMSProp")
print("Training Accuracy:",
      history_rmsprop.history["accuracy"][-1])
print("Validation Accuracy:",
      history_rmsprop.history["val_accuracy"][-1])
print("Training Loss:",
      history_rmsprop.history["loss"][-1])
print("Validation Loss:",
      history_rmsprop.history["val_loss"][-1])
print("Test Accuracy:",
      rmsprop_test_accuracy)
print("Test Loss:",
      rmsprop_test_loss)

print("\nAdam")
print("Training Accuracy:",
      history_adam.history["accuracy"][-1])
print("Validation Accuracy:",
      history_adam.history["val_accuracy"][-1])
print("Training Loss:",
      history_adam.history["loss"][-1])
print("Validation Loss:",
      history_adam.history["val_loss"][-1])
print("Test Accuracy:",
      adam_test_accuracy)
print("Test Loss:",
      adam_test_loss)

plt.figure(figsize=(10, 6))

plt.plot(
    history_sgd.history["val_loss"],
    label="SGD"
)

plt.plot(
    history_momentum.history["val_loss"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["val_loss"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["val_loss"],
    label="Adam"
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))

plt.plot(
    history_sgd.history["val_accuracy"],
    label="SGD"
)

plt.plot(
    history_momentum.history["val_accuracy"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["val_accuracy"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["val_accuracy"],
    label="Adam"
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

print("\nFINAL VALIDATION ACCURACY")

sgd_accuracy = history_sgd.history["val_accuracy"][-1]
momentum_accuracy = history_momentum.history["val_accuracy"][-1]
rmsprop_accuracy = history_rmsprop.history["val_accuracy"][-1]
adam_accuracy = history_adam.history["val_accuracy"][-1]

print("SGD:", sgd_accuracy)
print("Momentum:", momentum_accuracy)
print("RMSProp:", rmsprop_accuracy)
print("Adam:", adam_accuracy)

best_optimizer = max(
    [
        ("SGD", sgd_accuracy),
        ("Momentum", momentum_accuracy),
        ("RMSProp", rmsprop_accuracy),
        ("Adam", adam_accuracy)
    ],
    key=lambda x: x[1]
)

print("\nBest Optimizer:", best_optimizer[0])
print("Best Validation Accuracy:", best_optimizer[1])
```

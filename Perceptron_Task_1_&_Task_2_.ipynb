**Step 2**: Define the Perceptron class. It initializes the weights, computes the weighted sum, applies the activation function, predicts the output, updates the weights, and trains the model.

```python
import numpy as np

class Perceptron:

    def __init__(self, num_inputs, learning_rate=0.1):
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    def linear(self, inputs):
        return inputs @ self.weights[1:] + self.weights[0]

    def Heaviside_step_fn(self, z):
        return 1 if z >= 0 else 0

    def predict(self, inputs):
        Z = self.linear(inputs)

        if np.ndim(Z) == 0:
            return self.Heaviside_step_fn(Z)

        return [self.Heaviside_step_fn(z) for z in Z]

    def loss(self, prediction, target):
        return target - prediction

    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)

        self.weights[1:] += self.learning_rate * error * inputs
        self.weights[0] += self.learning_rate * error

    def fit(self, X, y, epochs):
        for epoch in range(epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)
```

**Step 3:** Create the AND gate dataset. It contains four input combinations and their expected outputs.

```python
X_and = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_and = np.array([0,0,0,1])
```

**Step 4**: Create the Perceptron model and train it using the AND gate dataset.

```python
model_and = Perceptron(num_inputs=2, learning_rate=0.1)
```

```python
model_and.fit(X_and, y_and, epochs=20)
```

**Step 5:** Predict the outputs of the AND gate using the trained Perceptron model.

**Step 6:** Calculate the accuracy by comparing the predicted outputs with the actual outputs.

```python
predictions = model_and.predict(X_and)
print(predictions)
```

```python
import numpy as np

class Perceptron:

    def __init__(self, num_inputs, learning_rate=0.1):
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    def linear(self, inputs):
        return inputs @ self.weights[1:] + self.weights[0]

    def Heaviside_step_fn(self, z):
        return 1 if z >= 0 else 0

    def predict(self, inputs):
        Z = self.linear(inputs)

        try:
            predictions = []
            for value in Z:
                predictions.append(self.Heaviside_step_fn(value))
            return predictions
        except TypeError:
            return self.Heaviside_step_fn(Z)

    def loss(self, prediction, target):
        return target - prediction

    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)

        self.weights[1:] += self.learning_rate * error * inputs
        self.weights[0] += self.learning_rate * error

    def fit(self, X, y, epochs):
        for epoch in range(epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)
```

**Step 4: **Separate the input features and target column. Convert categorical values into numerical values and prepare the dataset for training.



```python
# Separate features and target
X = df.drop("HeartDisease", axis=1)
y = df["HeartDisease"]

# Convert categorical columns
X = pd.get_dummies(X)

# Convert everything to numeric
X = X.apply(pd.to_numeric)

# Convert to NumPy float array
X = X.to_numpy(dtype=np.float64)

# Convert target to integer array
y = y.to_numpy(dtype=np.int64)
```

**Step 5: **Create the Single Layer Perceptron model by specifying the number of input features and learning rate.

```python
model = Perceptron(num_inputs=X.shape[1], learning_rate=0.01)
```

**Step 6:** Train the Perceptron model using the Heart Disease dataset for 100 epochs.

```python
model.fit(X, y, epochs=100)
```

**Step 7:** Predict the class labels for the input data using the trained model.

```python
predictions = model.predict(X)

print("First 10 Predictions:")
print(predictions[:10])
```

**Step 8:** Compare the predicted values with the actual values and calculate the accuracy of the model.

```python
accuracy = np.mean(np.array(predictions) == y)

print("Accuracy =", accuracy * 100, "%")
```

**Step 9:** Display the actual output and predicted output for the first 10 records.

```python
print("Actual vs Predicted")

for i in range(10):
    print("Actual:", y[i], " | Predicted:", predictions[i])
```

**Step 10:** Display the final weights learned by the Perceptron after training.

```python
print("Final Weights:")
print(model.weights)
```

**Conclusion:**

The Single Layer Perceptron was successfully trained using the Heart Disease Prediction dataset. The model learned the relationship between the input features and the target values, predicted the output, and achieved good classification accuracy.
